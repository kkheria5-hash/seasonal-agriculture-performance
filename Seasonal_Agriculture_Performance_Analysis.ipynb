## Economic and Risk Analysis

```python
season_profit_status=pd.crosstab(df['Season'],df['Profit_Status'],normalize='index')*100
display(season_profit_status.round(2))
season_profit_status.plot(kind='bar',figsize=(9,5)); plt.title('Profitability Percentage by Season'); plt.ylabel('Percentage (%)'); plt.xticks(rotation=0); plt.show()
```

## Statistical Analysis

```python
season_summary=df.groupby('Season').agg(Yield=('Yield_Tonnes_Ha','mean'),Production=('Production_Tonnes','mean'),Revenue=('Revenue_INR','mean'),Cost=('Total_Cost_INR','mean'),Profit=('Profit_INR','mean'),Water_Used=('Water_Used_m3','mean'),Water_Efficiency=('Water_Efficiency_t_per_1000m3','mean'),Disease_Risk=('Disease_Pest_Risk_pct','mean')).round(2)
display(season_summary)
```

```python
season_groups=[g['Yield_Tonnes_Ha'].values for _,g in df.groupby('Season')]
result=f_oneway(*season_groups)
print('ANOVA F-statistic:',round(result.statistic,4))
print('ANOVA p-value:',round(result.pvalue,6))
print('Significant difference across seasons.' if result.pvalue<0.05 else 'No statistically significant difference across seasons.')
```

```python
if {'Kharif','Rabi'}.issubset(set(df['Season'].unique())):
    a=df[df['Season']=='Kharif']['Yield_Tonnes_Ha']; b=df[df['Season']=='Rabi']['Yield_Tonnes_Ha']
    t,p=ttest_ind(a,b,equal_var=False)
    print('T-statistic:',round(t,4)); print('P-value:',round(p,6))
    print('Kharif and Rabi yields are significantly different.' if p<0.05 else 'No significant difference between Kharif and Rabi yields.')
else: print('Kharif and/or Rabi is not present in the dataset.')
```

## Key Findings

```python
best_season=df.groupby('Season')['Yield_Tonnes_Ha'].mean().idxmax()
worst_season=df.groupby('Season')['Yield_Tonnes_Ha'].mean().idxmin()
most_profitable=season_summary['Profit'].idxmax()
best_water=season_summary['Water_Efficiency'].idxmax()
print('Best season by yield:',best_season)
print('Worst season by yield:',worst_season)
print('Most profitable season:',most_profitable)
print('Most water-efficient season:',best_water)
print('Highest yielding crop:',crop_yield.idxmax())
print('Most profitable crop:',crop_profit.idxmax())
```

```python
final_comparison=df.groupby('Season').agg(Average_Yield=('Yield_Tonnes_Ha','mean'),Average_Production=('Production_Tonnes','mean'),Average_Revenue=('Revenue_INR','mean'),Average_Cost=('Total_Cost_INR','mean'),Average_Profit=('Profit_INR','mean'),Average_Water_Used=('Water_Used_m3','mean'),Average_Water_Efficiency=('Water_Efficiency_t_per_1000m3','mean'),Average_Disease_Risk=('Disease_Pest_Risk_pct','mean')).round(2)
display(final_comparison)
final_comparison.to_csv('seasonal_agriculture_summary.csv')
print('Summary saved as seasonal_agriculture_summary.csv')
```
